```python
s = 'date STRING , delay INT , distance INT , origin STRING , destination STRING'
```

```python
df =spark.read.csv('csv2/' , header = True , schema=s)
```

```python
df.cache()
```

```python
df.show(5)
```

```python
df.printSchema()
```

```python
# df.createOrReplaceTempView --> create a temprory view in memory , replace previous view with new one
```

```python
df.createOrReplaceTempView('us_flight_delays_view') # register as a view , view as a transformation
```

```python
# view craetion is lazy 
```

```python
df_short_fligts = spark.sql ("""
    SELECT * FROM us_flight_delays_view
    WHERE distance > 1000
""")
```

```python
df_short_fligts.show()
```

```python
from pyspark.sql.functions import col
```

```python
df_short_fligts = df.where(col('distance')>1000)
```

```python
df_short_fligts.show()
```

```python
df_short_fligts = spark.sql ("""
    SELECT * FROM us_flight_delays_view
    WHERE distance > 1000
    ORDER BY distance
""")
```

```python
df_short_fligts.show(5)
```

```python
df_short_fligts = spark.sql ("""
    SELECT * FROM us_flight_delays_view
    WHERE distance > 1000
    ORDER BY distance DESC
""")
```

```python
df_short_fligts.show()
```

```python
# Using SQL QUERY
#find all flights between SFO and ORD
#with delay  > 2 hours
```

```python
df_sfo_ord_delay = spark.sql ("""
    SELECT * From us_flight_delays_view
    WHERE origin in ('SFO' , 'ORD') AND destination in ('SFO' , 'ORD')
    AND delay > 120 
    ORDER BY delay DESC
""")
```

```python
df_sfo_ord_delay.show()
```

```python
df_new = spark.sql("""

    SELECT * ,
    CASE
        WHEN delay > 360 THEN 'very long delay'
        WHEN delay > 120 AND delay < 360 THEN 'long delay'
        WHEN delay > 60 AND delay < 120 THEN 'short delay'
        WHEN delay > 0 AND delay < 60 THEN 'tolerable delay'
        WHEN delay = 0 THEN 'no delay'
        ELSE 'Early'
        END AS Delay_Indecator
    FROM us_flight_delays_view 
""")
```

```python
df_new.show()
```

```python
df_new.select('Delay_Indecator').distinct().show()
```
